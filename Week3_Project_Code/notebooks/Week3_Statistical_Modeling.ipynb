Research question: Among records with complete numeric values, is Quantity associated with Total Spent after accounting for Price Per Unit?

```python
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import statsmodels.api as sm
from scipy import stats

data_path = Path("../data/retail_store_sales_raw.csv")
df = pd.read_csv(data_path)
for col in ["Price Per Unit", "Quantity", "Total Spent"]:
    df[col] = pd.to_numeric(df[col], errors="coerce")
print("Shape:", df.shape)
display(df.head())
display(df[["Price Per Unit", "Quantity", "Total Spent"]].describe().T)
print("Missing values:")
display(df[["Price Per Unit", "Quantity", "Total Spent"]].isna().sum())
```

## Hypotheses
- H₀: The coefficient for Quantity is zero after accounting for Price Per Unit.
- H₁: The coefficient for Quantity is not zero after accounting for Price Per Unit.

The model is a multiple linear regression. Since Total Spent is expected to be closely related to unit price × quantity, interpret results as a consistency/relationship check, not proof of causality.

```python
model_df = df[["Price Per Unit", "Quantity", "Total Spent"]].dropna().copy()
X = sm.add_constant(model_df[["Price Per Unit", "Quantity"]])
y = model_df["Total Spent"]
model = sm.OLS(y, X).fit()
print(model.summary())
print("\nQuantity coefficient:", model.params["Quantity"])
print("Quantity p-value:", model.pvalues["Quantity"])
print("R-squared:", model.rsquared)
```

## Residual diagnostics
Residual plots help identify patterns, unequal variance, and departures from model assumptions. A Q-Q plot provides a visual check of residual normality; it is not a definitive test.

```python
fitted = model.fittedvalues
residuals = model.resid
plt.figure(figsize=(7, 4.5))
plt.scatter(fitted, residuals, alpha=0.35)
plt.axhline(0, linestyle="--")
plt.xlabel("Fitted Total Spent")
plt.ylabel("Residual")
plt.title("Residuals vs Fitted Values")
plt.tight_layout()
plt.show()

sm.qqplot(residuals, line="45", fit=True)
plt.title("Q-Q Plot of Regression Residuals")
plt.tight_layout()
plt.show()
```

## Interpretation and limitations
Interpret coefficient estimates and p-values from the fitted output. R-squared measures the proportion of observed variation explained by the model in this sample. This is not a causal analysis; Total Spent is closely linked to price and quantity by construction. The analysis uses complete numeric rows, so results apply to that subset.
